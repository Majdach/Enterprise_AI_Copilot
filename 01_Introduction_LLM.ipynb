```python

from ollama import chat

def tester_temperature(temperature):
    response = chat(
        model="qwen2.5:3b",
        messages=[
            {
                "role": "system",
                "content": "Tu es un assistant créatif. Réponds en français."
            },
            {
                "role": "user",
                "content": "Propose une definition de l intelligence artificielle dans le contexte d une startup d'intelligence artificielle."
            }
        ],
        options={
            "temperature": temperature
        }
    )

    print(f"\nTempérature : {temperature}")
    print(response.message.content)
```

```python

tester_temperature(0)
tester_temperature(0.7)
tester_temperature(1.5)
```

```python

from ollama import chat

historique = [
    {
        "role": "system",
        "content": "Tu es un assistant RH professionnel. Réponds en français."
    }
]
```

```python

def discuter(question):
    # Ajouter la question de l'utilisateur
    historique.append({
        "role": "user",
        "content": question
    })

    # Envoyer tout l'historique à Qwen
    response = chat(
        model="qwen2.5:3b",
        messages=historique,
        options={"temperature": 0.2}
    )

    # Récupérer la réponse
    reponse = response.message.content

    # Sauvegarder la réponse dans l'historique
    historique.append({
        "role": "assistant",
        "content": reponse
    })

    print("Assistant :", reponse)
```

```python

discuter("Bonjour, je m'appelle Majda et je travaille au service RH.")
```

```python

discuter("Comment je m'appelle et dans quel service je travaille ?")
```

```python

print("Enterprise AI Copilot démarré !")
print("Tapez 'quitter' pour arrêter.\n")

while True:

    question = input("Vous : ")

    if question.lower().strip() == "quitter":
        print("Conversation terminée.")
        break

    if not question.strip():
        print("Veuillez écrire une question.")
        continue

    discuter(question)
```
